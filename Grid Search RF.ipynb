```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("50_startups_dataset.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset, drop_first=True)
```

```python
independent=dataset[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
dependent=dataset[['Profit']]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.2,random_state=11)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor

param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'bootstrap': [True, False]
}

grid = GridSearchCV(
    estimator=RandomForestRegressor(),
    param_grid=param_grid,
    refit=True,
    verbose=3,
    n_jobs=-1
)

grid.fit(x_train, y_train)
```

```python
regressor=grid.cv_results_
print("Best parameter{}:",format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(regressor)
```
